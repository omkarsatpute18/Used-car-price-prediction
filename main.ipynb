```python
#creating features and target
X = df.drop("selling_price",axis=1)
y = df['selling_price']
```

```python
X = X.drop("name",axis=1)
```

```python
#making categorical and numerical features
categorical_features = [
    "fuel",
    "seller_type",
    "transmission",
    "owner",
    "brand"
]

numerical_features = [
    "year",
    "km_driven"
]
```

```python
#importing some important sklearn features 
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
#Converting the categorical_features into numerical values

preprocessor = ColumnTransformer(
    transformers=[
        ("categotcal",OneHotEncoder(handle_unknown="ignore"),categorical_features)
    ],remainder="passthrough"
)
```

```python
#Splittinh the datset into 80:20 ratio 
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
#Our dataset is splitted into 80:20 ratio
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
#using the linearregrssion model
linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)
```

```python
#Training the linear regression model
linear_model.fit(X_train, y_train)
```

```python
y_pred_linear = linear_model.predict(X_test)
```

```python
#Lets evalute the linear regression model
mae_linear = mean_absolute_error(y_test, y_pred_linear)

rmse_linear = np.sqrt(
    mean_squared_error(y_test, y_pred_linear)
)

r2_linear = r2_score(y_test, y_pred_linear)

print("Linear Regression")

print("MAE :", mae_linear)
print("RMSE:", rmse_linear)
print("R²  :", r2_linear)
```

```python
#decision tree model
tree_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            DecisionTreeRegressor(
                max_depth=10,
                random_state=42
            )
        )
    ]
)
```

```python
#training our model on using decision tree model
tree_model.fit(X_train,y_train)
```

```python
#Prediction of decision tree
y_pred_tree = tree_model.predict(X_test)
```

```python
#evaluating the decision tree model
mae_tree = mean_absolute_error(y_test, y_pred_tree)

rmse_tree = np.sqrt(
    mean_squared_error(y_test, y_pred_tree)
)

r2_tree = r2_score(y_test, y_pred_tree)

print("Decision Tree")

print("MAE :", mae_tree)
print("RMSE:", rmse_tree)
print("R²  :", r2_tree)
```

```python
#Random forest model
rf_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
```

```python
#training our dataset using random forest
rf_model.fit(X_train, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test)
```

```python
#Evaluate the random forest
mae_rf = mean_absolute_error(y_test, y_pred_rf)

rmse_rf = np.sqrt(
    mean_squared_error(y_test, y_pred_rf)
)

r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest")

print("MAE :", mae_rf)
print("RMSE:", rmse_rf)
print("R²  :", r2_rf)
```

```python
#Compare the three models according to their values
results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],
    
    "MAE": [
        mae_linear,
        mae_tree,
        mae_rf
    ],
    
    "RMSE": [
        rmse_linear,
        rmse_tree,
        rmse_rf
    ],
    
    "R2 Score": [
        r2_linear,
        r2_tree,
        r2_rf
    ]
})

results
```

```python
#Visulaize the model
plt.figure(figsize=(10, 5))

sns.barplot(
    data=results,
    x="Model",
    y="R2 Score"
)

plt.title("Model Comparison Based on R² Score")
plt.ylim(0, 1)

plt.show()
```
